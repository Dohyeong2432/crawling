## 선별 발송 — 특정 게시글만 골라서 보내기

위 발송 셀들은 해당 사이트의 신규 자료를 **전부** 보냅니다.
일부만 보내려면 아래 셀의 `title_keywords` 에 제목을 넣으세요.

```python
###############################################################################
## [1단계] 무엇이 발송될지 먼저 확인 (메일 발송 안 함)
##
##  title_keywords 에 제목의 일부(또는 전체)를 넣습니다.
##   - 공백은 무시하고 비교하므로 사이트에서 제목을 복사해 붙여도 됩니다.
##   - 여러 개를 넣으면 그 중 하나라도 맞는 건이 대상입니다.
##   - 빈 목록([]) 이나 None 으로 두면 해당 사이트 전체가 대상이 됩니다.
###############################################################################
title_keywords = [
    '소비자 눈높이에 맞는 공모펀드',
]

## 확인할 사이트 (아래 중 하나를 고르세요)
site, site_name = fss_press, '금감원 보도자료'
# site, site_name = fsc_press, '금융위 보도자료'
# site, site_name = fiu_press, '금융정보분석원 보도자료'
# site, site_name = fss_itp,   '금융위 법령해석'

move_to_home(browser, {fss_itp:'fss_itp', fsc_press:'fsc_press',
                       fss_press:'fss_press', fiu_press:'fiu_press'}[site])
_result = site.get_table_data(browser)
_df = _result[0] if isinstance(_result, tuple) else _result
_df = _df[_df.등록일 >= today]
if title_keywords:
    _df = _df[_df['제목'].map(lambda s: subject_matches(s, title_keywords))]

print(f"[{site_name}] 발송 대상 {len(_df)}건   수신자: {', '.join(RECV_ADDRS)}")
display(_df[['제목', '등록일']] if len(_df) else '조건에 맞는 자료가 없습니다.')
```

```python
###############################################################################
## [2단계] 위에서 확인한 건만 실제로 발송
##  - 위 셀의 title_keywords / site 를 그대로 사용합니다.
##  - 이 셀을 실행하면 메일이 실제로 나갑니다. 되돌릴 수 없습니다.
###############################################################################
_notice = {fss_itp: fss_itp.notice_fss_ipt, fsc_press: fsc_press.notice_fsc_press,
           fss_press: fss_press.notice_fsc_press, fiu_press: fiu_press.notice_fiu_press}[site]
_notice(browser, today, title_keywords)
```
